```python
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()
```

Python gets confused because you are trying to read and change the local counter at the same time, so it gives an UnboundLocalError.

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
