```python
#Ques.27
import pandas as pd
import numpy as np

travel_df = pd.DataFrame({
    "Place": ["Delhi", "Agra", "Jaipur", "Mumbai", "Pune"]
})

# Add Code using np.arange(5)
travel_df["Code"] = np.arange(5)

print(travel_df)

# Create Series with values only at index 1 and 4
codes = pd.Series(
    ["A", "B"],
    index=[1, 4]
)

# Replace Code column
travel_df["Code"] = codes

print(travel_df)
```

```python
#Ques.28
import pandas as pd

data = {
    "Student": ["Rahul", "Aman", "Priya", "Neha"],
    "Class": [10, 10, 11, 11],
    "Section": ["A", "B", "A", "B"]
}

school_df = pd.DataFrame(data)

# Request a column that does not exist
new_df = school_df.reindex(
    columns=["Student", "Class", "Section", "House"]
)

# Assign one common value to complete House column
new_df["House"] = "Red"

print(new_df)
```

```python
#Ques.29
import pandas as pd

data = {
    "Item": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Quantity": [10, 20, 15, 8, 5]
}

inventory_df = pd.DataFrame(data)

# Partially indexed Series
stock = pd.Series(
    [100, 200],
    index=[1, 3]
)

# Add new column
inventory_df["Stock"] = stock

print(inventory_df)

# Delete Stock column
inventory_df.drop("Stock", axis=1, inplace=True)

print(inventory_df)
```

```python
#Ques.30
import pandas as pd

# First copy a table from Excel/Google Sheets
lab_df = pd.read_clipboard()

# 1. Display DataFrame
print(lab_df)

# 2. Display column names
print(lab_df.columns)

# 3. Retrieve one single-word column
print(lab_df["Name"])

# 4. Retrieve column whose name contains a space
print(lab_df["Total Marks"])

# 5. Create smaller DataFrame
small_df = lab_df[["Name", "Total Marks", "City"]]

print(small_df)

# 6. Add a new column
small_df["Batch"] = "2026"

print(small_df)

# 7. Delete the new column
small_df.drop("Batch", axis=1, inplace=True)

print(small_df)
```
