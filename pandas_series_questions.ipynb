```python
marks = {
    'Maths': 88,
    'Science': 74,
    'English': 69,
    'History': 81
}

marks_series = pd.Series(
    marks,
    index=['History', 'English', 'Science', 'Maths']
)

print("Complete Series:")
print(marks_series)

print("\nMarks greater than 70:")
print(marks_series[marks_series > 70])
```

```python
north = pd.Series({
    'Jan': 40,
    'Feb': 55,
    'Mar': 60
})

south = pd.Series({
    'Feb': 20,
    'Mar': 25,
    'Apr': 30
})

total = north + south

print("North:")
print(north)

print("\nSouth:")
print(south)

print("\nTotal:")
print(total)
```

```python
d = {
    'Red': 12,
    'Blue': 18,
    'Green': 24
}

labels = ['Green', 'Red', 'Yellow', 'Blue']

result = pd.Series(d, index=labels)

print("Result:")
print(result)

print("\nIs Null:")
print(pd.isnull(result))

print("\nYellow is NaN:", pd.isnull(result['Yellow']))

print("\nNumeric Values:")
print(result[pd.notnull(result)])
```

```python
s = pd.Series(
    [90, 75, 82],
    index=['Ravi', 'Aman', 'Neha']
)

d = s.to_dict()

s2 = pd.Series(d)

print("Original Series:")
print(s)

print("\nDictionary:")
print(d)

print("\nNew Series:")
print(s2)

print("\nSame labelled data:", s.equals(s2))
```

```python
products = pd.Series(
    [650, 450, 800, 300],
    index=['Laptop', 'Mouse', 'Phone', 'Keyboard']
)


print("Values:")
print(products.values)


print("\nIndex:")
print(products.index)


print("\nPhone Sales:")
print(products['Phone'])


print("\nSales above 500:")
print(products[products > 500])


print("\nIs Laptop present?")
print('Laptop' in products.index)
product_dict = products.to_dict()

print("\nDictionary:")
print(product_dict)
```

```python
data = {
    'A': 15,
    'B': 35,
    'C': 25
}

s = pd.Series(
    data,
    index=['C', 'A', 'D', 'B']
)
s.name = 'Observations'
s.index.name = 'Category'
print("Complete Series:")
print(s)

# (a)
print("\n(a) Index Order:")
print(s.index)

# (b)
print("\n(b) Label containing NaN:")
print(s[s.isnull()].index)

# (c)
print("\n(c) Values greater than 20:")
print(s[s > 20])

# (d)
print("\n(d) notnull():")
print(pd.notnull(s))


print("\n(e) Series Name:")
print(s.name)
print("Index Name:")
print(s.index.name)

```
